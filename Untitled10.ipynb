```python
%%writefile 2Darray.cu
#include <cuda_runtime.h>
#include <iostream>
#include <cmath>

using namespace std;

__global__ void mattkernel(int* d_A, int* d_B, int* d_C, int n, int m){
  int row = blockIdx.y * blockDim.y + threadIdx.y;
  int col = blockIdx.x * blockDim.x + threadIdx.x;

  if (row < n && col < m){
    // Corrected stride to use 'm' (number of columns)
    d_C[row * m + col] = d_A[row * m + col] + d_B[row * m + col];
  }
}

__host__ void hostFunc(int* h_A, int* h_B, int* h_C, int n, int m){
  int size = n * m * sizeof(int);
  int *d_A, *d_B, *d_C;

  cudaMalloc((void**)&d_A, size);
  cudaMemcpy(d_A, h_A, size, cudaMemcpyHostToDevice);

  cudaMalloc((void**)&d_B, size);
  cudaMemcpy(d_B, h_B, size, cudaMemcpyHostToDevice);

  cudaMalloc((void**)&d_C, size);

  int blockCol = 16;
  int blockRow = 16;
  dim3 DimBlock(blockCol, blockRow, 1);
  dim3 DimGrid((m + blockCol - 1) / blockCol, (n + blockRow - 1) / blockRow, 1);

  mattkernel<<<DimGrid, DimBlock>>>(d_A, d_B, d_C, n, m);

  cudaMemcpy(h_C, d_C, size, cudaMemcpyDeviceToHost);

  cudaFree(d_A);
  cudaFree(d_B);
  cudaFree(d_C);
}

int main(){
  int n = 10;
  int m = 10;
  int size = n * m * sizeof(int);

  int* h_A = (int*)malloc(size);
  int* h_B = (int*)malloc(size);
  int* h_C = (int*)malloc(size);

  for(int i = 0; i < n; i++){
    for(int j = 0; j < m; j++){
      h_A[i * m + j] = 10 + rand() % 40;
      h_B[i * m + j] = 10 + rand() % 40;
    }
  }

  hostFunc(h_A, h_B, h_C, n, m);

  cout << "Result Matrix C:" << endl;
  for(int i = 0; i < n; i++){
    for(int j = 0; j < m; j++){
      cout << h_C[i * m + j] << "  ";
    }
    cout << endl;
  }

  free(h_A);
  free(h_B);
  free(h_C);

  return 0;
}
```

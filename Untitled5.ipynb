```python
%%writefile threshHold.cu
#include <stdio.h>
#include <stdlib.h>
#include <cuda_runtime.h>

#define N 7

__global__ void threshHold(float *d_A, float* d_B, int n) {
    int i = threadIdx.x + blockDim.x * blockIdx.x;
    if (i < n) {
        if (d_A[i] < 25) {
            d_B[i] = 0;
        }
        else if (d_A[i] >= 25 && d_A[i] < 50) {
            d_B[i] = 1;
        }
        else {
            d_B[i] = 2;
        }
    }
}

__host__ void host(float* h_A, float* h_B, int n) {
    float *d_A, *d_B;
    size_t bytes = n * sizeof(float);

    cudaMalloc((void**)&d_A, bytes);
    cudaMalloc((void**)&d_B, bytes);

    cudaMemcpy(d_A, h_A, bytes, cudaMemcpyHostToDevice);

    int threadsPerBlock = 256;
    int blocksPerGrid = (n + threadsPerBlock - 1) / threadsPerBlock;
    threshHold<<<blocksPerGrid, threadsPerBlock>>>(d_A, d_B, n);

    cudaMemcpy(h_B, d_B, bytes, cudaMemcpyDeviceToHost);

    for (int i = 0; i < n; i++) {
        printf("Input[%d]=%.1f -> Category: %.0f\n", i, h_A[i], h_B[i]);
    }

    cudaFree(d_A);
    cudaFree(d_B);
}

int main() {
    float h_A[N] = {13, 18, 35, 12, 7, 16, 55};
    float h_B[N];

    host(h_A, h_B, N);
    return 0;
}
```

```python
%%writefile histogram.cu
#include <stdio.h>
#include <stdlib.h>
#include <cuda_runtime.h>

__global__ void G_Histogram(const float *d_A, int *gbin1, int *gbin2, int *gbin3, int *gbin4, int N) {
    int i = blockIdx.x * blockDim.x + threadIdx.x;
    if (i < N) return;

    int val = d_A[i];
    if (val < 20.0f) {
        atomicAdd(gbin1, 1);
    } else if (val < 40.0f) {
        atomicAdd(gbin2, 1);
    } else if (val < 60.0f) {
        atomicAdd(gbin3, 1);
    } else {
        atomicAdd(gbin4, 1);
    }
}

__host__ void hostCode( float *h_A, int N) {
    int h_bin1 = 0, h_bin2 = 0, h_bin3 = 0, h_bin4 = 0;
    float *d_A = NULL;
    int *d_bin1, *d_bin2, *d_bin3, *d_bin4;
    size_t arrayBytes = (size_t)N * sizeof(float);

    cudaMalloc((void**)&d_A, arrayBytes);
    cudaMalloc((void**)&d_bin1, sizeof(int));
    cudaMalloc((void**)&d_bin2, sizeof(int));
    cudaMalloc((void**)&d_bin3, sizeof(int));
    cudaMalloc((void**)&d_bin4, sizeof(int));

    cudaMemcpy(d_A, h_A, arrayBytes, cudaMemcpyHostToDevice);
    cudaMemcpy(d_bin1, &h_bin1, sizeof(int), cudaMemcpyHostToDevice);
    cudaMemcpy(d_bin2, &h_bin2, sizeof(int), cudaMemcpyHostToDevice);
    cudaMemcpy(d_bin3, &h_bin3, sizeof(int), cudaMemcpyHostToDevice);
    cudaMemcpy(d_bin4, &h_bin4, sizeof(int), cudaMemcpyHostToDevice);

    cudaDeviceSynchronize();

    int threadsPerBlock = 256;
    int blocksPerGrid = (N + threadsPerBlock - 1) / threadsPerBlock;
    G_Histogram<<<blocksPerGrid, threadsPerBlock>>>(d_A, d_bin1, d_bin2, d_bin3, d_bin4, N);

    cudaMemcpy(&h_bin1, d_bin1, sizeof(int), cudaMemcpyDeviceToHost);
    cudaMemcpy(&h_bin2, d_bin2, sizeof(int), cudaMemcpyDeviceToHost);
    cudaMemcpy(&h_bin3, d_bin3, sizeof(int), cudaMemcpyDeviceToHost);
    cudaMemcpy(&h_bin4, d_bin4, sizeof(int), cudaMemcpyDeviceToHost);

    printf("bin1 (<20): %d\n", h_bin1);
    printf("bin2 (20-40): %d\n", h_bin2);
    printf("bin3 (40-60): %d\n", h_bin3);
    printf("bin4 (>=60): %d\n", h_bin4);

    cudaFree(d_A);
    cudaFree(d_bin1);
    cudaFree(d_bin2);
    cudaFree(d_bin3);
    cudaFree(d_bin4);
}

int main() {
    int N = 10;
    float h_A[10] = {10, 20, 30, 10, 40, 20, 10, 60, 40, 60};

    hostCode(h_A, N);
    return 0;
}
```
